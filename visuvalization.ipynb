```python
import pandas as pd
df_sales=pd.read_excel("line.xlsx")
df_sales.head()
```

```python
from matplotlib import pyplot as plt
plt.plot(df_sales["Quarter"],df_sales["Fridge"])
```

```python
#the above output has extra line ([<matplotlib.lines.Line2D at 0x1efffd42510>])'from matplotlib import pyplot as plt' to remove use plt.show()
from matplotlib import pyplot as plt
plt.plot(df_sales["Quarter"],df_sales["Fridge"])
plt.show()
```

```python
#to change the width and height of the ploting plt.figure(figsize=(<width>,<height>))
#there are two function Figure and figure, use figure
from matplotlib import pyplot as plt
plt.figure(figsize=(14,7))
plt.plot(df_sales["Quarter"],df_sales["Fridge"])
plt.show()
```

```python
#now ploting revenue of others Dishwasher and washing machine
from matplotlib import pyplot as plt
plt.figure(figsize=(14,7))
plt.plot(df_sales["Quarter"],df_sales["Fridge"])
plt.plot(df_sales["Quarter"],df_sales["Dishwasher"],color="green")
plt.plot(df_sales["Quarter"],df_sales["Washing Machine"])
plt.show()
```

```python
#adding title,xlable, ylable and lable for each item also it is importatn for plt.legend()

from matplotlib import pyplot as plt
plt.figure(figsize=(14,7))
plt.plot(df_sales["Quarter"],df_sales["Fridge"], label="Fridge")
plt.plot(df_sales["Quarter"],df_sales["Dishwasher"],color="green", label="Dishwasher")
plt.plot(df_sales["Quarter"],df_sales["Washing Machine"], label="Washing machine")
plt.title("product sales")
plt.ylabel("revenue ( million $ )")
plt.xlabel("Financial Quater")

plt.legend()
plt.show()
```

```python
total_sales = df_sales[["Fridge", "Dishwasher","Washing Machine"]].sum()
```

```python
total_sales
```

```python
total_sales.index
```

```python
#piechart
plt.pie(total_sales, labels=total_sales.index)
```

```python
plt.pie(total_sales, labels=total_sales.index)
plt.show()
```

```python
#to get % in the pie use autopct key word
plt.pie(total_sales, labels=total_sales.index, autopct="%1.3f%%")
plt.show()
```

```python
#explode parameter for looking pie vizuvally good
plt.pie(total_sales, labels=total_sales.index, autopct="%1.3f%%",explode=(0.2,0,0), shadow=True)
plt.show()
```

```python
#explode parameter for looking pie vizuvally good
plt.pie(total_sales, labels=total_sales.index, autopct="%1.3f%%",explode=(0.2,0.3,0), shadow=True)
plt.show()
```

```python
#to change angle using startangle
plt.pie(total_sales, labels=total_sales.index, autopct="%1.3f%%",explode=(0.2,0,0), shadow=True,startangle=140)
plt.show()
```
