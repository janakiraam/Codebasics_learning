```python
import csv
with open("movies.csv") as f:
    data = list(csv.reader(f))
    print(data)
```

```python
header =  data[0]
data = data[1:]
```

```python
import csv

def calculate_rating_stats(data, industry=None):
    rating = []
    for row in data:
        if row[3]!='NULL' and ( not industry or row[1]== industry):
            rating.append(float(row[3]))

    max_rating=max(rating)
    min_rating=min(rating)
    avg_rating=sum(rating)/len(rating)
    return max_rating, min_rating,avg_rating

max_rating, min_rating, avg_rating=calculate_rating_stats(data)
print(f"for all movies the minmum rating is {min_rating}, maximum rating is {max_rating}, average rating is {avg_rating}")

max_rating, min_rating, avg_rating=calculate_rating_stats(data,industry="Bollywood")
print(f"for bollywood movies the minmum rating is {min_rating}, maximum rating is {max_rating}, average rating is {avg_rating}")

max_rating, min_rating, avg_rating=calculate_rating_stats(data,industry="Hollywood")
print(f"for hollywood movies the minmum rating is {min_rating}, maximum rating is {max_rating}, average rating is {avg_rating}")
```
