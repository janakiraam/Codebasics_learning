```python
f = open("funny.txt","r")
for i in f:
    print(i)

f.close()
```

```python
with open("funny.txt","r") as f:
    for line in f:
        print(line)


#f.close is not important as the context window will open at with open and it ends when the context window is closing.
```

```python
#executing every line in one shot

with open("funny.txt","r") as f:
    lines = f.readlines()
    print(lines)


#other modes then r -> read or w-> write and a -> append mode


with open("funny.txt","a") as f:
    f.write('\n teacher: what is this?')
```
